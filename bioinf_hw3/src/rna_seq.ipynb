## Работа с HISAT2

````r
%%bash
cd /content # перешла в нужную папку
all_samples=("SRR3414629" "SRR3414630" "SRR3414631" "SRR3414635" "SRR3414636" "SRR3414637")
for sample in "${all_samples[@]}";
do
    # -p 3 - количество потоков, -x mm10/genome - путь к индексу референсного генома, -U - Single-End чтения, -S - указывает выходной SAM файл
    # 2> ${sample}.hisat - перенаправление стандартного потока диагностики
    time hisat2 -p 3 -x mm10/genome -U ${sample}_1.fastq -S ${sample}_1.sam 2> ${sample}.hisat # запуск hisat2

    # grep - поиск по шаблону
    # -P -
    # '^@|NH:i:1$' - шаблон поиска:
    # ^@ - строки, начинающиеся с @ (заголовки SAM файла)
    # | - логическое или
    # NH:i:1$ - строки, заканчивающиеся на NH:i:1 (уникальные выравнивания)
    grep -P '^@|NH:i:1$' ${sample}_1.sam > ${sample}.uniq.sam # отбираю только те чтения, которые откартировались уникально (флаг NH:i:1)

    rm -v ${sample}_1.sam # удаляю исходные SAM файлы
done
````

## Статистика

````r
!grep '^_' SRR3414636.counts
````

````r
%%bash
all_samples=("SRR3414629" "SRR3414630" "SRR3414631" "SRR3414635" "SRR3414636" "SRR3414637")
sample_types=("Перепрограммированный" "Перепрограммированный" "Перепрограммированный" "Контрольный" "Контрольный" "Контрольный")

echo "ID | Тип образца | Общее кол-во исходных чтений | Кол-во и процент чтений, которые были успешно откартированы на геном | Кол-во и процент уникально откартированных чтений | Общее кол-во чтений, которые попали на гены"

for i in {0..5};
do
    sample_id="${all_samples[i]}"
    sample_type="${sample_types[i]}" # тип образца

    # awk '{print $1}' берет нужную колонку и возвращает значение из нее
    total_reads=$(grep "reads; of these" ${sample_id}.hisat | awk '{print $1}') # всего чтений

    # процент успешно откартированных чтений. tr -d '%' - удалить % из строки
    mapped_percent=$(grep "overall alignment rate" ${sample_id}.hisat | awk '{print $1}'| tr -d '%')
    # количество откартированных чтений. bc - выполнить математические вычисления
    mapped_reads=$(echo "scale=0; $total_reads * $mapped_percent / 100" | bc)

    # количество и процент уникально откартированных чтений
    uniquely_mapped=$(grep "aligned exactly 1 time" ${sample_id}.hisat | awk '{print $1}')
    uniquely_mapped_percent=$(grep "aligned exactly 1 time" ${sample_id}.hisat | awk '{print $2}' | tr -d '()%')

    # общее количество чтений, которые попали на гены (сумма всех генов, исключая __no_feature __ambiguous)
    reads_on_genes=$(grep -v "^__" ${sample_id}.counts | awk '{sum += $2} END {print sum}')

    # Форматируем вывод в виде таблицы
    echo "$sample_id | $sample_type | $total_reads | $mapped_reads ($mapped_percent%) | $uniquely_mapped ($uniquely_mapped_percent%) | $reads_on_genes"
done
````

````r
# установка BiocManager
if (!require("BiocManager", quietly = TRUE))
    install.packages("BiocManager")

# установка DESeq2
BiocManager::install("DESeq2")

# установка пакета для визуализации
install.packages("pheatmap")

# загрузка библиотек в текущую сессию
library(DESeq2)
library(pheatmap)
````

Создала файл ALL.info со следущим содержимым:

```
id condition
r1 SRR3414629 reprogramming
r2 SRR3414630 reprogramming
r3 SRR3414631 reprogramming
c1 SRR3414635 control
c2 SRR3414636 control
c3 SRR3414637 control
```

````r
# загрузка данных из ALL.info с информацией по каждому образцу
info   <- read.table("ALL.info", header=TRUE, sep="")

# загрузка данных об экспрессии. на выходе - матрица для DESeq2
# row.names=1 - первая колонка (с ensenbl id) станет названиями строк
counts <- as.matrix(read.table("ALL.counts", header=TRUE, row.names=1))

# cоздание объекта DESeqDataSet
# countData = counts - матрица с количеством ридов для каждого гена
# colData = info - данные, описывающие каждый образец
# design = ~ condition - cравниваnm экспрессию генов между группами, определенными в колонке condition
dds <- DESeqDataSetFromMatrix(countData = counts, colData = info, design = ~ condition)
````

````r
head(info)
````

````r
head(counts)
````

````r
# установка контрольной группы: группа control является контрольной(reference)
dds$condition <- relevel(dds$condition, ref="control")

# запуск анализа
# сравнивается reprogrammed и control для каждого гена
# для каждого гена считается: средняя экспрессия, изменение экспрессии, p-value
dds <- DESeq(dds)

# извлечение результатов
# alpha=0.01 - порог значимости для p-value
res <- results(dds, alpha=0.01)
````

## Heatmap

````r
# тепловая карта (heatmap) для визуализации сходства между образцами на основе экспрессии
rld <- rlog(dds) # стабилизация дисперсии данных
sampleDists <- dist(t(assay(rld))) # матрица расстояний
sampleDistMatrix <- as.matrix(sampleDists) # преобразование объекта "dist" в матрицу для построения heatmap
pheatmap(sampleDistMatrix, clustering_distance_rows=sampleDists, clustering_distance_cols=sampleDists) # построение heatmap
````

Все контрольные образцы похожи между собой, а перепрограммированные образцы - между собой. Такие кластеры доказывают присутствие дифференицальной экспрессии и хорошее качество эксперимента.

## plotMA

````r
# Установка и загрузка biomaRt (для того, чтобы получить названия генов по ENSEBLE ID)
if (!require("biomaRt", quietly = TRUE)) {
    BiocManager::install("biomaRt")
}
library(biomaRt)
````

````r
# cортировка результатов по p-value (padj)
# самые значимые - вверху списка
resOrdered <- res[order(res$padj),]

# отбор статистически значимых генов
resSignificant <- subset(resOrdered, padj < 0.01)

# подключение к базе данных мыши
ensembl <- useMart("ensembl", dataset = "mmusculus_gene_ensembl")

# получение ENSEMBL ID (убираю версии .1, .2 ...)
ensembl_ids <- gsub("\\..*", "", rownames(resSignificant))

# получение названий генов
gene_names <- getBM(
    attributes = c("ensembl_gene_id", "external_gene_name"),
    filters = "ensembl_gene_id",
    values = ensembl_ids,
    mart = ensembl
)

# аннотированная таблица
resSignificant_annotated <- as.data.frame(resSignificant)
resSignificant_annotated$gene_name <- gene_names$external_gene_name[match(ensembl_ids, gene_names$ensembl_gene_id)]

# название гена в первой колонке
resSignificant_annotated <- resSignificant_annotated[, c("gene_name", "baseMean", "log2FoldChange", "lfcSE", "stat", "pvalue", "padj")]

# сохранение значимых генов в файл
# sep="\t" - табуляция как разделитель
# row.names=TRUE - сохранять названия генов (ENSEMBL ID)
# quote=FALSE - не заключать значения в кавычки
write.table(resSignificant_annotated, "differentially_expressed_genes.txt", sep="\t", row.names=TRUE, quote=FALSE)
````

Синим цветом показаны те гены, в которых значительно изменился уровень экспрессии. Серым цветом показаны гены, в которых не произошло статистически значимых измений уровня экспрессии. Из графика видно, что множество генов значимо изменили экспрессию, наблюдаются сильные изменения - log2FC до 10 (изменение в 1000+ раз)

## Тепловая карта для 20 самых экспрессированных генов

````r
# counts(dds, normalized=TRUE) - нормализованные counts
# rowMeans() - средняя экспрессия для каждого гена по всем образцам
# order(..., decreasing=TRUE) - сортировка по убыванию экспрессии
# [1:20] - 20 самых экспрессированных генов
select <- order(rowMeans(counts(dds, normalized=TRUE)), decreasing=TRUE)[1:20]

# colData(dds) - информация об образцах из DESeqDataSet
# [,c("condition","id")] - только колонки condition и id
df <- as.data.frame(colData(dds)[,c("condition","id")])

# построение тепловой карты:
pheatmap(assay(rld)[select,],
         cluster_rows=FALSE,
         show_rownames=FALSE,
         cluster_cols=FALSE,
         annotation_col=df)
````

Из графика видно, что гены равномерно экспрессированы во всех образцах, нет разделения на контроль и перепрограммирование. Это подтверждает хорошее качество данных, так как наиболее экспрессированные гены связаны с базовыми процессами, и их экспрессия отличается слабо.
