## Извлечение Pfam ID из таблицы

````python
import csv
import re

# Список для хранения выбранных семейств
families = []

epigenetics = ["Chromatin remodeling cofactor", "Chromatin remodeling", "Histone modification read",
"Histone modification write cofactor", "Histone modification read", "Histone modification", "Histone chaperone"]

with open('For project - selected.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)  # читаем как словарь

    for row in reader:
        if row['Modification'] == '#' or row['Function'] not in epigenetics:
            continue

        symbol = row['HGNC approved symbol']  # Берём название семейства
        if symbol == 'AEBP2': # неспецифичен, даёт 1900+ хитов
                      continue
        pfam_str = row['Pfam domains'] # Берём строку с Pfam доменами
        pfam_ids = re.findall(r'PF\d{5}', pfam_str) # Ищем все PFxxxxx

        # Если нашлись Pfam домены, добавляем семейство в список
        if pfam_ids:
            families.append({
                'family': symbol,
                'pfam_domains': ', '.join(pfam_ids),
                'function': row['Function'],
                'modification': row['Modification']
            })

        # выход, если набрали 10 семейств
        if len(families) == 10:
            break

print("10 семейств: ")
for i, f in enumerate(families, 1):
    print(f"{i}. {f['family']}")
    print(f"   Pfam: {f['pfam_domains']}")
    print(f"   Function: {f['function']}")
    print(f"   Modification: {f['modification']}")
    print()

# все Pfam ID из всех семейств в один список
all_pfam_ids = []
for f in families:    all_pfam_ids.extend(f['pfam_domains'].split(', '))

# удаление дубликатов
unique_pfam_ids = list(dict.fromkeys(all_pfam_ids))

print(f"Уникальных Pfam доменов для поиска: {len(unique_pfam_ids)}")
print(unique_pfam_ids)
````

## Hmmer

````python
!pip install pyhmmer pandas
````

````python
from pyhmmer import hmmscan, easel, plan7
import pandas as pd

pfam_list = unique_pfam_ids

# загрузка базы Pfam
with plan7.HMMFile("Pfam-A.hmm") as hmm_file:
    all_hmm = list(hmm_file)

hmm_needed = []
for hmm in all_hmm:
    acc = hmm.accession
    acc_clean = acc.split('.')[0]  # PF10417.12 в PF10417
    if acc_clean in pfam_list:
        hmm_needed.append(hmm)

print(f"Найдено {len(hmm_needed)} HMM из {len(pfam_list)}")

# загрузка протеома осьминога
with easel.SequenceFile("protein.faa", digital=True) as sf:
    sequences = list(sf)

print(f"Загружено {len(sequences)} белков")

results = hmmscan(sequences, hmm_needed, cpus=2)
hits = []
for top_hits in results:
    protein = top_hits.query.name
    for hit in top_hits:
        for domain in hit.domains:
            # Берём Accession и отрезаем версию
            acc_full = domain.alignment.hmm_accession
            if isinstance(acc_full, bytes):
                acc_full = acc_full.decode()
            pfam_id = acc_full.split('.')[0]   # PFxxxxx
            hits.append({
                'pfam_domain': pfam_id,
                'protein_id': protein,
                'score': round(domain.score, 2),
                'evalue': f'{domain.c_evalue:.2e}'
            })

df = pd.DataFrame(hits).drop_duplicates()
df = df[(df['evalue'].astype(float) < 1e-10) & (df['score'] > 25)]
df.to_csv('results.csv', index=False)

print(f"Найдено {len(df)} совпадений")
print(df.head())
````

````python
import re
from Bio import SeqIO
import pandas as pd

# Файлы
input_fasta = "/content/GCA_951406725.2_xcOctVulg1.2_genomic.fna"
output_csv = "g4_quadruplexes.csv"

# Паттерн для G4
pattern = re.compile(r"(G{3,5}[ATGC]{1,7}){3,}G{3,5}", re.IGNORECASE)

# Функция реверс-комплемента
def revcomp(seq):
    return seq[::-1].translate(str.maketrans("ATGC", "TACG"))

hits = []

for record in SeqIO.parse(input_fasta, "fasta"):
    chrom = record.id
    seq = str(record.seq).upper()
    rev = revcomp(seq)
    print(f"Ищу в {chrom}...")

    # + стренд
    for m in pattern.finditer(seq):
        hits.append([chrom, m.start(), m.end(), '+', m.group()])

    # - стренд
    for m in pattern.finditer(rev):
        start = len(seq) - m.end()
        end = len(seq) - m.start()
        hits.append([chrom, start, end, '-', m.group()])

df = pd.DataFrame(hits, columns=['chrom', 'start', 'end', 'strand', 'seq'])
df.to_csv(output_csv, index=False)

print(f"Найдено {len(df)} G4. Результат в {output_csv}")
````

````python
import pandas as pd
import numpy as np
from intervaltree import IntervalTree
from collections import defaultdict
````

````python
# Для g4 (квадруплексы) – start и end уже есть, сделаем end полуоткрытым
g4['start_half'] = g4['start']
g4['end_half'] = g4['end'] + 1

# Для zdnabert – добавим start и end (включительные)
zdnabert['start_half'] = zdnabert['POSITION']
zdnabert['end_half'] = zdnabert['POSITION'] + zdnabert['LENGTH'] + 1

# Для zhunt – аналогично
zhunt['start_half'] = zhunt['POSITION']
zhunt['end_half'] = zhunt['POSITION'] + zhunt['LENGTH'] + 1

# Для аннотации – start и end уже есть, переведём в полуоткрытые
annotations_df['start_half'] = annotations_df['start']
annotations_df['end_half'] = annotations_df['end'] + 1
````

````python
# Возвращает долю пересечения feature с region: (0,1)
def overlap_ratio(region_start, region_end, feat_start, feat_end):
    inter_start = max(region_start, feat_start)
    inter_end = min(region_end, feat_end)
    if inter_start >= inter_end:
        return 0.0
    inter_len = inter_end - inter_start
    feat_len = feat_end - feat_start
    return inter_len / feat_len
````

````python
# Построение интервальных деревьев
# Принимает DataFrame с предсказаниями. Возвращает словарь: {seqid: IntervalTree}, где каждый интервал хранит индекс строки
def build_interval_trees(df, seqid_col='seqid', start_col='start_half', end_col='end_half'):
    trees = defaultdict(IntervalTree)
    for idx, row in df.iterrows():
        seqid = row[seqid_col]
        # Используем полуоткрытые интервалы [start, end)
        trees[seqid].addi(row[start_col], row[end_col], idx)
    return trees

# Построим деревья для всех трёх наборов
g4_trees = build_interval_trees(g4, seqid_col='chrom')
zhun_trees = build_interval_trees(zhunt, seqid_col='seqid')
zdnabert_trees = build_interval_trees(zdnabert, seqid_col='seqid')
````

````python
# Создание датафрейма с регионами
regions_list = []

# Вспомогательная функция для обрезки отрицательных координат
def clip_negative(start, end):
    return max(start, 0), end

# Экзоны
exons = annotations_df[annotations_df['type'] == 'exon'][['seqid', 'start_half', 'end_half']].copy()
for _, row in exons.iterrows():
    regions_list.append([row['seqid'], row['start_half'], row['end_half'], 'Exons'])

# Гены (нужны для промоторов, downstream, интронов)
genes = annotations_df[annotations_df['type'] == 'gene'][['seqid', 'start_half', 'end_half', 'strand']].copy()

# Промоторы (1000 bp вверх от TSS)
for _, row in genes.iterrows():
    if row['strand'] == '+':
        start = row['start_half'] - 1000
        end = row['start_half']
        start, end = clip_negative(start, end)
    else:  # strand == '-'
        start = row['end_half']
        end = row['end_half'] + 1000
    regions_list.append([row['seqid'], start, end, 'Promoters'])

# Downstream (200 bp после гена)
for _, row in genes.iterrows():
    if row['strand'] == '+':
        start = row['end_half']
        end = row['end_half'] + 200
    else:
        start = row['start_half'] - 200
        end = row['start_half']
        start, end = clip_negative(start, end)
    regions_list.append([row['seqid'], start, end, 'Downstream'])

# Интроны
for _, gene in genes.iterrows():

    exon_mask = (exons['seqid'] == gene['seqid']) & \
                (exons['start_half'] >= gene['start_half']) & \
                (exons['end_half'] <= gene['end_half'])
    gene_exons = exons[exon_mask].sort_values('start_half')
    if len(gene_exons) < 2:
        continue
    for i in range(len(gene_exons) - 1):
        intron_start = gene_exons.iloc[i]['end_half']
        intron_end = gene_exons.iloc[i+1]['start_half']
        if intron_start < intron_end:
            regions_list.append([gene['seqid'], intron_start, intron_end, 'Introns'])

# Межгенные области
for seqid in genes['seqid'].unique():
    genes_on_chr = genes[genes['seqid'] == seqid].sort_values('start_half')
    if len(genes_on_chr) < 2:
        continue
    for i in range(len(genes_on_chr) - 1):
        gap_start = genes_on_chr.iloc[i]['end_half']
        gap_end = genes_on_chr.iloc[i+1]['start_half']
        if gap_start < gap_end:
            regions_list.append([seqid, gap_start, gap_end, 'Intergenic'])

# Создаём единый DataFrame
regions_df = pd.DataFrame(regions_list, columns=['seqid', 'start_half', 'end_half', 'region_type'])
````

````python
print(f"Всего регионов: {len(regions_df)}")
regions_df.head()
````

````python
# интервальные деревья для каждого типа региона

region_trees = {}
for region_type in regions_df['region_type'].unique():
    subset = regions_df[regions_df['region_type'] == region_type]
    trees = defaultdict(IntervalTree)
    for _, row in subset.iterrows():
        trees[row['seqid']].addi(row['start_half'], row['end_half'], None)
    region_trees[region_type] = trees
````

````python
# Подсчёт уникальных структур, попавших в регионы

def count_structures_in_regions(struct_trees, region_trees, overlap_threshold=0.9):
    results = {}
    for region_type, r_trees in region_trees.items():
        structure_ids = set()
        for seqid, rtree in r_trees.items():
            if seqid not in struct_trees:
                continue
            stree = struct_trees[seqid]
            for s_interval in stree:
                s_start, s_end, s_id = s_interval.begin, s_interval.end, s_interval.data
                overlapping = rtree.overlap(s_start, s_end)
                for r_interval in overlapping:
                    if overlap_ratio(r_interval.begin, r_interval.end, s_start, s_end) >= overlap_threshold:
                        structure_ids.add(s_id)
                        break
        results[region_type] = len(structure_ids)
    return results
````

````python
# Подсчёт регионов, содержащих хотя бы одну структуру

def count_regions_with_structures(region_trees, struct_trees, overlap_threshold=0.9):
    results = {}
    for region_type, r_trees in region_trees.items():
        region_counter = 0
        for seqid, rtree in r_trees.items():
            if seqid not in struct_trees:
                continue
            stree = struct_trees[seqid]
            for r_interval in rtree:
                r_start, r_end = r_interval.begin, r_interval.end
                overlapping = stree.overlap(r_start, r_end)
                for s_interval in overlapping:
                    if overlap_ratio(r_start, r_end, s_interval.begin, s_interval.end) >= overlap_threshold:
                        region_counter += 1
                        break
        results[region_type] = region_counter
    return results
````

````python
# Вычисление результатов

# Общее количество структур каждого типа
total_g4 = len(g4)
total_zhun = len(zhunt)
total_zdnabert = len(zdnabert)

# Первая таблица: число структур в регионах
g4_counts = count_structures_in_regions(g4_trees, region_trees)
zhun_counts = count_structures_in_regions(zhun_trees, region_trees)
zdnabert_counts = count_structures_in_regions(zdnabert_trees, region_trees)

# Вторая таблица: число регионов со структурами
regions_with_g4 = count_regions_with_structures(region_trees, g4_trees)
regions_with_zhun = count_regions_with_structures(region_trees, zhun_trees)
regions_with_zdnabert = count_regions_with_structures(region_trees, zdnabert_trees)

# Общее количество регионов каждого типа (из regions_df)
total_regions = regions_df.groupby('region_type').size().to_dict()

for rt in region_trees.keys():
    if rt not in total_regions:
        total_regions[rt] = 0
````

````python
# Таблица 1
table1_data = []
for region_type in sorted(region_trees.keys()):
    table1_data.append([
        region_type,
        g4_counts[region_type],
        f"{g4_counts[region_type]/total_g4:.2%}" if total_g4 else "0%",
        zhun_counts[region_type],
        f"{zhun_counts[region_type]/total_zhun:.2%}" if total_zhun else "0%",
        zdnabert_counts[region_type],
        f"{zdnabert_counts[region_type]/total_zdnabert:.2%}" if total_zdnabert else "0%"
    ])
table1 = pd.DataFrame(table1_data, columns=[
    'Участок', 'Число квадруплексов', 'Доля квадруплексов',
    'Число предсказаний Zhun', 'Доля предсказаний Zhun',
    'Число предсказаний ZDNABERT', 'Доля предсказаний ZDNABERT'
])
print("Таблица 1:")
print(table1.to_string(index=False))

# Таблица 2
table2_data = []
for region_type in sorted(region_trees.keys()):
    total = total_regions.get(region_type, 0)
    table2_data.append([
        region_type,
        regions_with_g4[region_type],
        f"{regions_with_g4[region_type]/total:.2%}" if total else "0%",
        regions_with_zhun[region_type],
        f"{regions_with_zhun[region_type]/total:.2%}" if total else "0%",
        regions_with_zdnabert[region_type],
        f"{regions_with_zdnabert[region_type]/total:.2%}" if total else "0%"
    ])
table2 = pd.DataFrame(table2_data, columns=[
    'Участок', 'Число участков с квадруплексом', 'Доля участков с квадруплексом',
    'Число участков с предсказанием Zhun', 'Доля участков с предсказанием Zhun',
    'Число участков с предсказанием ZDNABERT', 'Доля участков с предсказанием ZDNABERT'
])
print("\nТаблица 2:")
print(table2.to_string(index=False))
````

````python
def show_hits(filename, name):
    print(f"\n===== {name} =====")
    try:
        with open(filename, 'r') as f:
            lines = [l for l in f if not l.startswith('#') and l.strip()]
        if not lines:
            print("Нет попаданий (файл пуст)")
            return
        data = []
        for line in lines:
            cols = line.split()
            if len(cols) >= 6 and float(cols[4]) < 1e-5:
                data.append([cols[0], cols[4], cols[5]])
        if data:
            df = pd.DataFrame(data, columns=['Protein_ID', 'E-value', 'Score'])
            print(df.to_string(index=False))
        else:
            print("Нет белков с E-value < 1e-5")
    except FileNotFoundError:
        print("Файл не найден")

show_hits('smc_results.txt', 'SMC (когезин)')
show_hits('rad21_results.txt', 'RAD21 (когезин)')
show_hits('c2h2_results.txt', 'CTCF (цинковые пальцы PF00096)')
show_hits('ctcf_results.txt', 'CTCF (N-term PF01132)')
````

## Поиск генов, в промоторах которых есть zdna

````python
# Берём гены из аннотации
genes = annotations_df[annotations_df['type'] == 'gene'].copy()

def get_gene_name(attrs):
    for part in attrs.split(';'):
        if part.startswith('Name='):
            return part.split('=')[1]
    return None

genes['gene_name'] = genes['attributes'].apply(get_gene_name)
genes['gene_id'] = genes['attributes'].str.extract(r'ID=([^;]+)')[0]

# Строим промоторы (1000 bp выше TSS)
promoters_list = []
for _, gene in genes.iterrows():
    if gene['strand'] == '+':
        start = max(0, gene['start_half'] - 1000)
        end = gene['start_half']
    else:
        start = gene['end_half']
        end = gene['end_half'] + 1000
    promoters_list.append({
        'seqid': gene['seqid'],
        'start': start,
        'end': end,
        'gene_id': gene['gene_id'],
        'gene_name': gene['gene_name']
    })
promoters_df = pd.DataFrame(promoters_list)
````

````python
from collections import defaultdict

genes_with_zdna = set()

for _, prom in promoters_df.iterrows():
    seqid = prom['seqid']
    if seqid not in zdnabert_trees:
        continue
    overlaps = zdnabert_trees[seqid].overlap(prom['start'], prom['end'])
    if overlaps:
        genes_with_zdna.add((prom['gene_id'], prom['gene_name']))

print(f"Найдено {len(genes_with_zdna)} генов с Z‑ДНК в промоторах")
````

````python
import re

prot_to_gene = {}
with open("bimac_genomic.gff", "r") as f:
    for line in f:
        if line.startswith("#"):
            continue
        parts = line.strip().split("\t")
        if len(parts) < 9 or parts[2] != "CDS":
            continue
        attrs = parts[8]
        # Ищем protein_id=XP_...
        p_match = re.search(r"protein_id=([^;]+)", attrs)
        if not p_match:
            continue
        prot_id = p_match.group(1)
        # Ищем gene=LOC... (или просто gene=...)
        g_match = re.search(r"gene=([^;]+)", attrs)
        if g_match:
            gene_id = g_match.group(1)   # например, LOC106877895
            prot_to_gene[prot_id] = gene_id

print(f"Найдено соответствий: {len(prot_to_gene)}")
# Применяем к вашему blast_df
blast_df['bimac_gene_id'] = blast_df['hit_id_bimac'].map(prot_to_gene)
print(blast_df[['gene_id', 'hit_id_bimac', 'bimac_gene_id']].head(10))
````

````python
from collections import defaultdict

def count_structures_in_region_trees(region_trees, struct_trees):
    """
    region_trees: dict {region_type: {seqid: IntervalTree}}
    struct_trees: dict {seqid: IntervalTree}
    Возвращает dict {region_type: количество структур}
    """
    counts = defaultdict(int)
    for region_type, r_trees in region_trees.items():
        for seqid, rtree in r_trees.items():
            if seqid not in struct_trees:
                continue
            stree = struct_trees[seqid]
            # Для каждого интервала региона ищем пересечения со структурами
            for r_interval in rtree:
                overlaps = stree.overlap(r_interval.begin, r_interval.end)
                counts[region_type] += len(overlaps)
    return counts

real_g4 = count_structures_in_region_trees(region_trees, g4_trees)
real_zhun = count_structures_in_region_trees(region_trees, zhun_trees)
real_zdnabert = count_structures_in_region_trees(region_trees, zdnabert_trees)
````

````python
region_lengths = regions_df.groupby('region_type').apply(
    lambda x: (x['end_half'] - x['start_half']).sum()
).to_dict()
total_length = sum(region_lengths.values())
````

````python
total_g4 = len(g4)
total_zhun = len(zhunt)
total_zdnabert = len(zdnabert)

expected_g4 = {r: (l / total_length) * total_g4 for r, l in region_lengths.items()}
expected_zhun = {r: (l / total_length) * total_zhun for r, l in region_lengths.items()}
expected_zdnabert = {r: (l / total_length) * total_zdnabert for r, l in region_lengths.items()}
````

````python
enrich_g4 = {r: real_g4.get(r, 0) / expected_g4[r] for r in region_lengths}
enrich_zhun = {r: real_zhun.get(r, 0) / expected_zhun[r] for r in region_lengths}
enrich_zdnabert = {r: real_zdnabert.get(r, 0) / expected_zdnabert[r] for r in region_lengths}
````

````python
import pandas as pd

rows = []
for region in region_lengths:
    rows.append({
        'Region': region,
        'Length (bp)': region_lengths[region],
        'G4 real': real_g4.get(region, 0),
        'G4 expected': round(expected_g4[region], 1),
        'G4 enrichment': round(enrich_g4[region], 2),
        'Zhun real': real_zhun.get(region, 0),
        'Zhun expected': round(expected_zhun[region], 1),
        'Zhun enrichment': round(enrich_zhun[region], 2),
        'ZDNABERT real': real_zdnabert.get(region, 0),
        'ZDNABERT expected': round(expected_zdnabert[region], 1),
        'ZDNABERT enrichment': round(enrich_zdnabert[region], 2),
    })

df_compare = pd.DataFrame(rows)
print(df_compare.to_string(index=False))
````
