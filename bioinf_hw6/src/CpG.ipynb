### Подсчет картированных ридов в заданных регионах

```python
%%bash

# функция для подсчета картированных ридов в заданных регионах
count_reads_in_regions() {
    local bam_file=$1

    # имя для отсортированного файла (без расширения .bam)
    local base_name=$(basename "$bam_file" .bam)
    local sorted_bam="${base_name}_sorted.bam"

    # сортировка bam-файла по хромосомным координатам и сохранение результата в новый файл sorted_bam
    samtools sort "$bam_file" -o "$sorted_bam"

    # индексация отсортированного файла
    samtools index "$sorted_bam"

    # подсчет для региона 1: 11347700-11367700
    count1=$(samtools view "$sorted_bam" chr11:11347700-11367700 | cut -f1 | sort -u | wc -l)

    # подсчет для региона 2: 40185800-40195800
    count2=$(samtools view "$sorted_bam" chr11:40185800-40195800 | cut -f1 | sort -u | wc -l)

    # возвращаем значения для таблицы
    echo "$base_name,$count1,$count2"
}


# обрабатка каждого bam-файла
for bam in *.bam; do
    if [ -f "$bam" ]; then
        result=$(count_reads_in_regions "$bam")
        echo "$result"
    fi
done
```

## Гистограммы распределения метилирования цитозинов по хромосоме

```python
import pandas as pd
import matplotlib.pyplot as plt
import gzip

# функция для построения гистограммы
def plot_methylation_histogram(cov_file, sample_name, coverage_threshold=10):
    """
    cov_file: путь к файлу .cov.gz
    sample_name: название образца для заголовка
    coverage_threshold: минимальная глубина покрытия для фильтрации
    """

    # чтение файла
    with gzip.open(cov_file, 'rt') as f:
        df = pd.read_csv(f, sep='\t', header=None,
                         names=['chr', 'start', 'end', 'percent', 'methylated', 'unmethylated'])

    # расчет глубины покрытия
    df['coverage'] = df['methylated'] + df['unmethylated']

    # статистика до фильтрации
    total_sites = len(df)
    print(f"\n{sample_name}:")
    print(f"всего позиций: {total_sites}")
    print(f"средняя глубина покрытия: {df['coverage'].mean():.2f}")

    # фильтрация по глубине
    filtered = df[df['coverage'] >= coverage_threshold]
    filtered_sites = len(filtered)
    print(f"позиций с покрытием >= {coverage_threshold}: {filtered_sites} ({100*filtered_sites/total_sites:.1f}%)")

    # построение гистограммы
    plt.figure(figsize=(10, 6))
    plt.hist(filtered['percent'], bins=50, edgecolor='black', alpha=0.7, color='steelblue')
    plt.xlabel('процент метилирования (%)')
    plt.ylabel('количество позиций (частота)')
    plt.title(f'{sample_name}\nРаспределение метилирования (покрытие >= {coverage_threshold}, n = {filtered_sites})')
    plt.grid(True, alpha=0.3)
    plt.show()


# вызов функции для трех образцов
plot_methylation_histogram('SRR5836473_1_dedup.deduplicated.bismark.cov.gz', '8-cell')
plot_methylation_histogram('SRR5836475_1_dedup.deduplicated.bismark.cov.gz', 'ICM')
plot_methylation_histogram('SRR3824222_1_dedup.deduplicated.bismark.cov.gz', 'Epiblast')
```

```python
#фильтрация по p-value и q-value

import pandas as pd

def filter_dmr(file_in, file_out, p=0.05):
    """фильтрация DMR по p-value"""
    df = pd.read_csv(file_in, sep='\t', header=None,
                     names=['chr', 'start', 'end', 'col4', 'mean_diff', 'num_cpgs', 'p_mwu', 'qval', 'mean_g1', 'mean_g2'])

    filtered = df[df['qval'] < p].dropna(subset=['qval'])
    filtered.to_csv(file_out, sep='\t', index=False, header=False)
    print(f"{file_in}: было {len(df)}, стало {len(filtered)} DMR")

def filter_dmc(file_in, file_out, p=0.05):
    """фильтрация DMC по p-value"""
    df = pd.read_csv(file_in, sep='\t', header=None,
                     names=['chr', 'start', 'end', 'col4', 'diff', 'col6', 'pval', 'col8', 'col9', 'col10'])
    filtered = df[df['pval'] < p].dropna(subset=['pval'])
    filtered.to_csv(file_out, sep='\t', index=False, header=False)
    print(f"{file_in}: было {len(df)}, стало {len(filtered)} DMС")

# фильтрация DMR
filter_dmr('dmr_8cell_vs_ICM_Epiblast.bed', 'dmr_8cell_vs_ICM_Epiblast_filt.bed')
filter_dmr('dmr_ICM_vs_8cell_Epiblast.bed', 'dmr_ICM_vs_8cell_Epiblast_filt.bed')
filter_dmr('dmr_Epiblast_vs_8cell_ICM.bed', 'dmr_Epiblast_vs_8cell_ICM_filt.bed')

# фильтрация DMC
filter_dmc('dmc_8cell_vs_ICM_Epiblast.bed', 'dmc_8cell_vs_ICM_Epiblast_filt.bed')
filter_dmc('dmc_ICM_vs_8cell_Epiblast.bed', 'dmc_ICM_vs_8cell_Epiblast_filt.bed')
filter_dmc('dmc_Epiblast_vs_8cell_ICM.bed', 'dmc_Epiblast_vs_8cell_ICM_filt.bed')
```

## В промоторы каких генов попадают DMR

```python
# скачиваю гены mm10 в формате BED
!wget https://hgdownload.cse.ucsc.edu/goldenPath/mm10/database/refGene.txt.gz
!gunzip refGene.txt.gz
```

```python
# загрузка генов
genes = pd.read_csv('refGene.txt', sep='\t', header=None,
                    names=['bin', 'name', 'chrom', 'strand', 'txStart', 'txEnd', 'cdsStart', 'cdsEnd',
                           'exonCount', 'exonStarts', 'exonEnds', 'score', 'name2', 'cdsStartStat',
                           'cdsEndStat', 'exonFrames'])
```

```python
# только хромосома 11 и нужные колонки
genes = genes[genes['chrom'] == 'chr11'][['chrom', 'txStart', 'txEnd', 'name2', 'strand']].copy()
genes.columns = ['chr', 'start', 'end', 'gene', 'strand']
```

```python
# убираю chr из названия хромосомы (для соответствия)
genes['chr'] = genes['chr'].str.replace('chr', '')
```

```python
# добавление промоторов
# upstream 1000 bp, downstream 100 bp
genes['prom_start'] = genes.apply(lambda x: x['start'] - 1000 if x['strand'] == '+' else x['end'] - 100, axis=1)
genes['prom_end'] = genes.apply(lambda x: x['start'] + 100 if x['strand'] == '+' else x['end'] + 1000, axis=1)
```

```python
def get_genes_from_dmr(dmr_file, q=0.05):
    """находит гены на chr11, чьи промоторы пересекаются с DMR"""

    # загрузка DMR
    dmr = pd.read_csv(dmr_file, sep='\t', header=None,
                      names=['chr', 'start', 'end', 'col4', 'mean_diff',
                             'n_cpg', 'p_mwu', 'qval', 'mean_g1', 'mean_g2'])

    # фильтрация по q-value
    dmr_filt = dmr[dmr['qval'] < q]

    # поиск пересечений с промоторами
    gene_list = []

    for _, dmr_row in dmr_filt.iterrows():
        # убираю chr из названия хромосомы
        dmr_chr = dmr_row['chr'].replace('chr', '')

        for _, gene_row in genes.iterrows():
            if dmr_chr == gene_row['chr']:
                # проверка пересечения DMR с промотором
                if (dmr_row['start'] < gene_row['prom_end'] and
                    dmr_row['end'] > gene_row['prom_start']):
                    gene_list.append(gene_row['gene'])

    unique_genes = list(set(gene_list))
    print(f"  Количество генов: {len(unique_genes)}")

    return unique_genes
```

```python
genes_8cell = get_genes_from_dmr("dmr_8cell_vs_ICM_Epiblast.bed")
```

```python
genes_icm = get_genes_from_dmr("dmr_ICM_vs_8cell_Epiblast.bed")
```

```python
genes_epi = get_genes_from_dmr("dmr_Epiblast_vs_8cell_ICM.bed")
```

```python
# объединение всех уникальных гены
all_genes = list(set(genes_8cell + genes_icm + genes_epi))
```

```python
!pip install gprofiler-official

import gprofiler
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# инициализация g:Profiler
gp = gprofiler.GProfiler(return_dataframe=True)

# GO-анализ

try:
    results = gp.profile(
        organism='mmusculus', # мышь
        query=all_genes,
        sources=['GO:BP'],  # биологические процессы
        user_threshold=0.05   # порог
    )

    # сортировка по p-value
    results = results.sort_values('p_value')

    # сохраняем
    results.to_csv('GO_results_all_genes.csv', index=False)

    print(f"\nнайдено {len(results)} значимых GO терминов")
    print("\nтоп-10 терминов:")
    print(results[['name', 'p_value', 'intersection_size']].head(10))

except Exception as e:
    print(f"ошибка: {e}")
```
