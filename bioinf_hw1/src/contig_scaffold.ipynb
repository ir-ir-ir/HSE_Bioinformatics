```python
from Bio import SeqIO
import numpy as np

contigs = [] # длины контигов
total_length = 0 # общая длина
n_50 = 0 # N50

for record in SeqIO.parse('Contigi_contig.fa', "fasta"):
    length = len(record.seq) # длина текущего контига
    contigs.append(length)   # добавляю в список
    total_length += length   # считаю общую длину

# сортирую по убыванию
contigs_sorted = sorted(contigs, reverse=True)

# расчет N50
half = total_length / 2
length_counting = 0

# попорядку перебираю длины контигов (от длинных к коротким)
for i in contigs_sorted:
  length_counting += i
  if length_counting >= half: # если достигли половины или превысили ее
    n_50 = i # записываю N50
    break

# Вывод результатов
print("Общее количество контигов: ", len(contigs))
print("Общая длина: ", total_length)
print("Длина самого длинного контига: ", contigs_sorted[0])
print("Длина самого короткого контига: ", contigs_sorted[-1])
print("Средняя длина контига: ", np.mean(contigs))
print("N50: ", n_50)
```

```python
from Bio import SeqIO
import numpy as np

scaffolds = [] # длины скаффолдов
total_length = 0 # общая длина
n_50 = 0 # N50

for record in SeqIO.parse('scaffolds_scaffold.fa', "fasta"):
    length = len(record.seq) # длина текущего скаффолда
    scaffolds.append(length) # добавляю в список
    total_length += length   # считаю общую длину

# сортирую по убыванию
scaffolds_sorted = sorted(scaffolds, reverse=True)

# расчет N50
half = total_length / 2
length_counting = 0

# попорядку перебираю длины скаффолдов (от длинных к коротким)
for i in scaffolds_sorted:
  length_counting += i
  if length_counting >= half: # если достигли половины или превысили ее
    n_50 = i # записываю N50
    break

# Вывод результатов
print("Общее количество скаффолдов: ", len(scaffolds))
print("Общая длина: ", total_length)
print("Длина самого длинного скаффолда: ", scaffolds_sorted[0])
print("Длина самого короткого скаффолда: ", scaffolds_sorted[-1])
print("Средняя длина скаффолда: ", np.mean(scaffolds))
print("N50: ", n_50)
```

```python
from Bio import SeqIO

# нахожу самый длинный скаффолд
longest_seq = ""
for record in SeqIO.parse('scaffolds_scaffold.fa', "fasta"):
    if len(record.seq) > len(longest_seq):
        longest_seq = str(record.seq) # преобразую в строку и запоминаю

print("Длина самого длинного скаффолда: ", len(longest_seq))

# подсчет гэпов
gap_count = 0
total_gap_length = 0
in_gap = False # булевая переменная - в гэпе ли ?

for i in longest_seq:
    if i == 'N':
        total_gap_length += 1
        if not in_gap:
            gap_count += 1
            in_gap = True
    else:
        in_gap = False

print("Количество гэпов: ", gap_count)
print("Общая длина гэпов: ", total_gap_length)
```
