# Сравнение результатов

```python
# набор пиков ChIP-seq из базы данных ENCODE
!wget https://www.encodeproject.org/files/ENCFF080LBO/@@download/ENCFF080LBO.bed.gz
```

```python
!gzip -d ENCFF080LBO.bed.gz
```

```python
import matplotlib
matplotlib.use('agg')                    # неинтерактивный бэкенд для matplotlib
import matplotlib.pyplot as plt          # для построения графиков
from matplotlib_venn import venn2        # для диаграмм Венна
import subprocess                        # для выполнения команд в терминале
import os                                # для работы с путями к файлам
```

```python
!pip install matplotlib-venn -q
```

```python
# отсавляю только нужную хромосому
!awk '$1 == "chr20"' ENCFF080LBO.bed > ENCFF080LBO_chr20.bed
```

```python
# Данные
files = ["macs3_IQS/NA_peaks.broadPeak", "macs3_VDZ/NA_peaks.broadPeak"]
name = ["/IQS.jpg","/VDZ.jpg"]
i = 0
encode_file = "ENCFF080LBO_chr20.bed"

for file in files:
  # Подсчёт пиков (каждая строка = один пик)
  my_peaks = sum(1 for _ in open(file))
  encode_peaks = sum(1 for _ in open(encode_file))

  # Подсчёт пересечения
  # выполнение команды bedtools в терминале
  # -a : первый файл (полученные пики)
  # -b : второй файл (ENCODE пики)
  # -u : выводить только уникальные пересечения
  # wc -l : подсчитывает количество строк (пиков)
  result = subprocess.run(f"bedtools intersect -a {file} -b {encode_file} -u | wc -l",
                            shell=True, capture_output=True, text=True)
  # преобразование вывода в число
  overlap = int(result.stdout.strip())

  # Построение диаграммы
  # subsets = (только a, только b, пересечение)
  plt.figure(figsize=(10, 6))
  venn2(subsets=(my_peaks - overlap, encode_peaks - overlap, overlap),
          set_labels=(f'My peaks\n({os.path.basename(file)})',
                      f'ENCODE peaks\n({os.path.basename(encode_file)})'))
  plt.title("Пересечение ChIP-seq пиков")

  # Создание папки и сохранение
  os.makedirs("venn_results", exist_ok=True)
  path = "venn_results" + name[i]
  plt.savefig(path, dpi=150, bbox_inches='tight')
  plt.show()

  percent_of_encode = (overlap / encode_peaks) * 100 if encode_peaks else 0

  print(f"{name[i]}:")
  print(f"Статистика")
  print(f"  Полученные пики:         {my_peaks}")
  print(f"  ENCODE пики:             {encode_peaks}")
  print(f"  Общие пики:              {overlap}")
  print(f"  % от ENCODE пиков:       {percent_of_encode:.1f}%\n")

  i += 1
```

Выводы:
1. VDZ имеет значительно больше собственных пиков (5806 против 1260 у IQS). Так как пиков больше, то и пересечений с ENCODE нашлось больше: 1815 против 263
2. В VDZ нашлось больше пиков. Объяснить это можно тем, что в реплике VDZ риды сконцентрированы в регионах энхансеров (создавая высокий сигнал относительно фона). В реплике IQS сигнал слабее, поэтому пиков найдено мало, несмотря на большое количество уникальных ридов. В IQS, вероятно, плохое антитело или ошибки в протоколе (технические ошибки).
3. Пики отличаются от эталонных ENCODE, потому что Peak Calling проводился только на одной хромосоме (chr20), что не позволило алгоритму MACS3 корректно оценить глобальный уровень шума.

```python
# файл с размерами хромосом
!wget -q https://hgdownload-test.gi.ucsc.edu/goldenPath/hg38/bigZips/hg38.chrom.sizes
# утилита bedGraphToBigWig
!wget -q http://hgdownload.soe.ucsc.edu/admin/exe/linux.x86_64/bedGraphToBigWig
!chmod 777 bedGraphToBigWig
```

```python
# скачиваю bam файлы (38 сборка)
!wget https://www.encodeproject.org/files/ENCFF579UIC/@@download/ENCFF579UIC.bam
!wget https://www.encodeproject.org/files/ENCFF733BZR/@@download/ENCFF733BZR.bam
```

```python
def bam_to_bigwig(bam_file, output_name):
    # BAM в BED
    !bedtools bamtobed -i {bam_file} > {output_name}.bed
    # только 20 хромосома
    !awk '$1 == "chr20"' {output_name}.bed > {output_name}_chr20.bed
    # сортировка BED
    # -i - входной BED файл
    # -k1,1 - сортировка по первому полю (хромосома)
    !sort -k1,1 {output_name}_chr20.bed > {output_name}_chr20_sorted.bed
    # BED в bedGraph
    # -i - входной BED файл
    # -g - файл с размерами хромосом
    # -bg - выход в формате bedGraph
    !bedtools genomecov -i {output_name}_chr20_sorted.bed -g hg38.chrom.sizes -bg > {output_name}.bedGraph
    # bedGraph в BigWig
    !./bedGraphToBigWig {output_name}.bedGraph hg38.chrom.sizes {output_name}.bigWig
    print(f"Файл {output_name}.bigWig создан.")
```

```python
bam_to_bigwig("ENCFF579UIC.bam", "ENCFF579UIC")
bam_to_bigwig("ENCFF733BZR.bam", "ENCFF733BZR")
```
